```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv(r"C:\Users\gowri\OneDrive\MLAssignment\df_train_test.csv")
```

```python
df.columns
```

### Distribution of Marital Status

```python
col = "marital_status"

plt.figure(figsize=(10,6))
ax = sns.countplot(data=df, x=col, order=df[col].value_counts().index, palette="Set2")
plt.title("Distribution of marital_status")
plt.xticks(rotation=45)

total = len(df[col])

# Add count + percentage labels
for p in ax.patches:
    count = int(p.get_height())
    percent = (count / total) * 100
    ax.annotate(
        f'{count} ({percent:.0f}%)',
        (p.get_x() + p.get_width() / 2, p.get_height()),
        ha='center', va='bottom'
    )

plt.tight_layout()
plt.show()
```

### Distribution of Occupation

```python
col = "occupation"

# Frequency tables
counts = df[col].value_counts()
print(counts)
print("\nProportions:")
print(df[col].value_counts(normalize=True))

plt.figure(figsize=(10,6))
ax = sns.countplot(
    data=df,
    y=col,
    order=counts.index,
    palette="Set1"
)
plt.title("Distribution of occupation")

total = len(df[col])
max_count = counts.max()

# Expand x‑axis so labels fit comfortably
plt.xlim(0, max_count * 1.15)

# Add count + percentage labels INSIDE the bar
for p in ax.patches:
    count = int(p.get_width())
    percent = (count / total) * 100

    # Position text slightly before the end of the bar
    ax.annotate(
        f'{count} ({percent:.0f}%)',
        (count * 0.98, p.get_y() + p.get_height() / 2),
        ha='right', va='center', color='white', fontsize=9, fontweight='bold'
    )

plt.tight_layout()
plt.show()
```
